```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 

# 1. Create a dummy dataset using NumPy
# Generating 500 data points from a normal distribution
np.random.seed(42)  # Setting seed for reproducibility
data = np.random.normal(loc=50, scale=10, size=500)

# 2. Convert the data into a Pandas DataFrame
df = pd.DataFrame(data, columns=["Values"])

# 3. Calculate statistical measures using Pandas and NumPy
mean_val = df["Values"].mean()
median_val = df["Values"].median()
variance_val = df["Values"].var()  # Sample variance by default
std_dev_val = df["Values"].std()  # Sample standard deviation by default

# Print the calculated metrics
print(f"--- Statistical Summary ---")
print(f"Mean:               {mean_val:.4f}")
print(f"Median:             {median_val:.4f}")
print(f"Variance:           {variance_val:.4f}")
print(f"Standard Deviation: {std_dev_val:.4f}")

# 4. Plot the graph using Matplotlib
plt.figure(figsize=(10, 6))

# Plot a histogram of the data
plt.hist(
    df["Values"], bins=30, edgecolor="black", color="skyblue", alpha=0.7
)

# Add vertical lines to point out Mean and Median
plt.axvline(
    mean_val,
    color="red",
    linestyle="dashed",
    linewidth=2,
    label=f"Mean ({mean_val:.2f})",
)
plt.axvline(
    median_val,
    color="green",
    linestyle="dotted",
    linewidth=2,
    label=f"Median ({median_val:.2f})",
)

# Customize the chart labels, titles, and legend
plt.title("Data Distribution with Mean and Median", fontsize=14)
plt.xlabel("Value Range", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend(loc="upper right")
plt.grid(axis="y", alpha=0.3)

# Display the plot
plt.show()
```
